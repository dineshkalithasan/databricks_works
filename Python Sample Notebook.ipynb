# Python Sample Notebook

A quick demonstration of common Python patterns: variables, data structures, a function, and a small pandas DataFrame.



```python
# 1. Variables and basic data structures
name = "Databricks"
numbers = [1, 2, 3, 4, 5]
squares = {n: n**2 for n in numbers}

print(f"Hello from {name}!")
print("Squares:", squares)

# 2. A simple function with a loop and conditional
def fizzbuzz(n: int) -> str:
    if n % 15 == 0:
        return "FizzBuzz"
    elif n % 3 == 0:
        return "Fizz"
    elif n % 5 == 0:
        return "Buzz"
    return str(n)

for i in range(1, 16):
    print(fizzbuzz(i), end=" " )
print()

# 3. A small pandas DataFrame
import pandas as pd

df = pd.DataFrame({
    "product": ["Laptop", "Tablet", "Monitor", "Keyboard"],
    "price": [1200.00, 450.50, 310.99, 79.25],
    "units_sold": [12, 34, 8, 55],
})

# Add a derived column
df["revenue"] = df["price"] * df["units_sold"]

display(df)
```
